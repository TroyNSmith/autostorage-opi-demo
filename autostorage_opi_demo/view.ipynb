```python
from pathlib import Path

import py3Dmol

path = Path("/home/tns97255/autostorage-opi-demo/out/4_NEB/optts.xyz")

# Initialize the py3Dmol viewer
view = py3Dmol.view(width=600, height=400)

# Add all frames to the viewer
view.addModelsAsFrames(path.read_text().replace(">\n", ""), "xyz")

# Set the visualization style for all models/frames
view.setStyle({}, {"stick": {"radius": 0.15}, "sphere": {"scale": 0.25}})

# Configure background and framing
view.setBackgroundColor("white")
view.zoomTo()

# Start the animation loop ('forward' or 'loop')
view.animate({"loop": "forward"})

view.show()
```
